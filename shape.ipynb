```python
import numpy as np

# Exercise 1 — shapes (3,) and (3,): trivial, same shape
a = np.array([1, 2, 3])
b = np.array([10, 20, 30])
print(a + b,'3')           # what do you expect before running?

# Exercise 2 — shapes (3,) and (1,): b stretches
a = np.array([1, 2, 3])
b = np.array([10])
print(a + b,'2')           # predict first

# Exercise 3 — shapes (3, 1) and (3,): this is the key one
# (3,1) + (3,) → (3,1) + (1,3) → (3,3)
col = np.array([[1], [2], [3]])    # shape (3, 1)
row = np.array([10, 20, 30])       # shape (3,)
print(col + row,'1')  
```
