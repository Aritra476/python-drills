```python
from astropy import constants as const

c = const.c.si.value
h = const.h.si.value
print(c,h)

c = const.c.to('km/s').value
h = const.h.si.value
print(c,h)
```

```python
from astropy import constants as const

c1 = const.c.to("km/s").value
c2 = const.c.to("m/s").value

print(f"speed of light in km/s={c1:.10e}")
print(f"speed of light in m/s={c2:.6f}")

#radial velocity
v = 30 #km/s
B = v/c1 
print(f"beta={B:.4e}")
```
